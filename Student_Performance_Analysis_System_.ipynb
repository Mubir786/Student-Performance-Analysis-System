```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

1. We will load our csv file

```python
df = pd.read_csv('/content/student_performance.csv')
df
```

Printing First Five students

```python
df.head(5)
```

Printing Last Five Students

```python
df.tail(5)
```

Calculating Mean

```python
subjects = ["Python", "Database", "Networking"]
mean = df[subjects].mean()
print(mean)
```

Calculating Maximum

```python
maximum = df[subjects].max()
print(maximum)
```

Calculating Minimum

```python
minimum = df[subjects].min()
print(minimum)
```

Calculating total marks of each student

```python
df["total"] = df[subjects].sum(axis=1)
df
```

Calculating percentage of each student

```python
df["Percentage"] = df[subjects].sum(axis=1)
df
```

Student with highest Percentage

```python
highest_percentage = df.loc[df['Percentage'].idxmax()]
highest_percentage
```

Pass if marks are at least 50 in every subject

```python
df["Status"] = np.where(
    (df[subjects] >= 50).all(axis=1),
    "Pass",
    "Fail"
)
```

```python
df["Status"].value_counts()
```

Plotting histogram

```python
plt.figure(figsize=(8, 5))
```

```python
plt.hist(df["Python"], bins=5, color="skyblue", edgecolor="black")
plt.title("Distribution of Python Marks")
plt.xlabel("Python Marks")
plt.ylabel("Number of Students")

plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```

Plotting Pie Chart

```python
average_marks = df[subjects].mean()

plt.figure(figsize=(7, 7))

plt.pie(
    average_marks,
    labels=average_marks.index,
    autopct="%1.1f%%",
    colors=["red", "green", "blue"],
    startangle=90
)

plt.title("Average Marks by Subject")

plt.tight_layout()
plt.show()
```

Calculating subject with highest average

```python
print(average_marks.idxmax())
print("Average:", round(average_marks.max(), 2))
```

Plotting Line Chart

```python
first_ten = df.head(10)

plt.figure(figsize=(11, 6))

plt.plot(
    first_ten["Student"],
    first_ten["Python"],
    color="red",
    marker="o",
    label="Python"
)
```

```python
plt.plot(
    first_ten["Student"],
    first_ten["Database"],
    color="green",
    marker="s",
    label="Database"
)
```

```python
plt.plot(
    first_ten["Student"],
    first_ten["Networking"],
    color="blue",
    marker="^",
    label="Networking"
)
```

```python
first_ten = df.head(10)

plt.figure(figsize=(11, 6))

plt.plot(
    first_ten["Student"],
    first_ten["Python"],
    color="red",
    marker="o",
    label="Python"
)

plt.plot(
    first_ten["Student"],
    first_ten["Database"],
    color="green",
    marker="s",
    label="Database"
)

plt.plot(
    first_ten["Student"],
    first_ten["Networking"],
    color="blue",
    marker="^",
    label="Networking"
)

plt.title("Marks of the First 10 Students")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.xticks(rotation=45)

plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

Calculating Student with highest Networking marks among first ten

```python
networking_top = first_ten.loc[
    first_ten["Networking"].idxmax()
]

print("\nHIGHEST NETWORKING MARKS AMONG FIRST TEN")
print("Student:", networking_top["Student"])
print("Marks:", networking_top["Networking"])
```
